```python
def average_multiple(L: List[int], k: int)-> float:
    """
    Returns the average of all numbers in L which are a multiple of k.
    If no numbers in L are a multiple of k it returns 0.0. Assume k > 0.
    Assume all values of L are greater than 0.
    >>> average_multiple([10, 31, 20, 15, 17], 2)
    15.0
    >>> average_multiple([1, 2, 3, 4, 6, 7, 8, 9], 3)
    6.0
    >>> average_multiple([1, 2, 3, 4], 1)
    2.5
    >>> average_multiple([1, 2, 3, 4], 5)
    0.0
    >>> average_multiple([], 1)
    0.0
    """
    # factors = [n for n in range(1,k) if k % n == 0]
    # print(factors)
    total = 0
    count = 0
    for element in L:
        if element % k == 0:
            total += element
            count+=1
    if count == 0:
        return 0
    return total/count

print(average_multiple([10, 31, 20, 15, 17], 2))
```

```python
def zero_all(values, to_zero)-> int:
    """
    Replaces any numbers in values which also appear in to_zero with 0.
    See the below for some examples.
    
    >>> L = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
    >>> zero_all(L, [1, 4, 9])
    >>> L
    L = [[0, 2, 3], [0, 5, 6], [7, 8, 0]]
    >>> L = [[1, 2, 2], [1, 5], []]
    >>> zero_all(L, [2, 9])
    >>> L
    L = [[1, 0, 0], [1, 5], []]
    >>> L = [[1, 2, 3], [], [7, 8, 9]]
    >>> zero_all(L, [])
    >>> L
    [[1, 2, 3], [], [7, 8, 9]]
    """
    for element in values:
        for i in range(len(element)):
            if element[i] in to_zero:
                element[i] = 0
    return values

L = [[1, 2, 2], [1, 5], []]
zero_all(L, [2, 9])
```

```python
def delete_all(s:str, to_delete:str)-> str:
    new_string = ''
    for char in s:
        if char not in to_delete:
            new_string += char
    return new_string

print( delete_all("Hello world!", "lo"))
```

```python
def count_teams(teams: List[List[int]], student)-> int:
    count = 0
    for element in teams:
        for j in range(len(element)):
            if element[j] == student:
                count +=1
    return count

print(count_teams([[1,2,3],[1,5,6],[1]], 1))
print(count_teams([[1,2,3],[1,5,6],[1]], 4))
print(count_teams([[1,2,3],[1,5,6],[1]], 5))
print(count_teams([], 8))
```

```python
def is_valid_number(num):
    if num.count('.') == 0:
        return num.isnumeric()
    n = num.index('.')
    return num[:n].isnumeric() and num[n:].isnumeric()

is_valid_number("abc")
```

```python
def two_sum(arr,target):
    num_to_index = {}
    for  i,n in enumerate(arr):
        complement = target - n
        if complement in num_to_index:
            return [num_to_index[complement],i]
        num_to_index[n] = i


print(two_sum([2,7,11,15],9))
```

```python
def two_pointer(word):
    l,r = 0, len(word)-1
    while l< r: 
        while l < r and not word[l].isalnum():
            l+=1
        while l < r and  not word[r].isalnum():
            r-=1
        if word[l].lower() != word[r].lower():
            return False
        l+=1
        r-=1
    return True

print(two_pointer("tacocat"))
```
